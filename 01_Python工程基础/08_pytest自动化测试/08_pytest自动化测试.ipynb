## 1. 为什么需要自动化测试

以前我们经常这样检查代码：

```python
print(add(2, 3))
```

然后用眼睛判断输出是不是5。项目变大后，几十个功能不可能每次都手动检查。

自动化测试会让计算机替我们比较实际结果与期望结果。

````python
def add(left: int, right: int) -> int:
    return left + right


assert add(2, 3) == 5
print("断言通过，所以程序继续执行")
````

`assert 条件`表示：我认为这个条件必须成立。

- 条件为`True`：什么也不发生；
- 条件为`False`：抛出`AssertionError`。

测试框架会收集这些成功和失败结果。

## 2. pytest怎样找到测试

pytest默认识别：

- 名字以`test_`开头的文件；
- 名字以`test_`开头的函数。

例如：

```python
# test_calculator.py
def test_add():
    assert add(2, 3) == 5
```

在终端执行：

```powershell
python -m pytest -v
```

`-v`表示显示更详细的测试名称。

## 3. 一个测试的三个部分

常用结构是Arrange、Act、Assert：

```python
def test_discount():
    prices = [100, 50]                         # Arrange：准备
    actual = calculate_total(prices, 0.2)     # Act：执行
    assert actual == 120.0                    # Assert：验证
```

简单测试可以合并写，但思考时仍要分清准备、执行和验证。

## 4. 测试异常

如果规则要求负价格抛出`ValueError`，测试应该验证异常类型：

```python
def test_rejects_negative_price():
    with pytest.raises(ValueError):
        calculate_total([10, -1])
```

含义是：缩进代码必须抛出`ValueError`。如果没有异常，或者抛出其他异常，测试失败。

## 5. 参数化测试

同一个规则需要测试多组数据时，可以使用`pytest.mark.parametrize`：

```python
@pytest.mark.parametrize(
    ("value", "expected"),
    [(1, 2), (2, 4), (3, 6)],
)
def test_double(value, expected):
    assert value * 2 == expected
```

pytest会把它当作三次独立测试运行。

## 6. 怎样阅读测试失败

测试失败时先看：

1. 哪个测试函数失败；
2. 实际值和期望值分别是什么；
3. 异常类型是否与预期一致；
4. Traceback最后指向自己的哪一行代码。

不要看到红色就立刻修改测试。测试表达需求时，应该修正业务代码；只有测试本身写错时才修改测试。
