## 1. 什么是模块

简单来说，一个 Python 文件就是一个模块。例如同目录中的 `calculator.py` 是文件，导入后则是一个模块对象。

模块可以集中保存函数、类和常量，避免把所有代码写进一个文件。

````python
import calculator

print(calculator)
print(calculator.add(2, 3))
print(calculator.PI)
````

导入整个模块后，通过 `模块名.成员名` 使用其中的内容。这样写通常最清楚，因为读者一眼就知道 `add` 来自哪里。

````python
from calculator import add, circle_area

print(add(10, 20))
print(circle_area(2))
````

`from calculator import add` 会把 `add` 这个名字直接放进当前环境，因此调用时不用写 `calculator.add()`。

暂时不要使用 `from calculator import *`，因为它会把很多名字放进当前环境，难以判断名字来自哪里。

## 2. `__name__` 是什么

每个 Python 模块都有一个 `__name__` 变量：

- 文件被直接运行时，`__name__ == "__main__"`；
- 文件被导入时，`__name__` 通常是模块名。

因此下面的入口判断可以让某些代码只在直接运行文件时执行：

```python
if __name__ == "__main__":
    main()
```

刚才导入 `calculator` 时，没有看到“calculator.py 被直接运行”，就是因为入口条件不成立。

````python
print("当前 Notebook 中的 __name__：", __name__)
print("calculator 模块中的 __name__：", calculator.__name__)
````
