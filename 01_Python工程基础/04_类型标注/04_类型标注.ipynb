## 1. 参数与返回值标注

```python
def add(left: int, right: int) -> int:
    return left + right
```

`left: int`和`right: int`描述参数类型，`-> int`描述返回值类型。

````python
def add(left: int, right: int) -> int:
    return left + right


print(add(2, 3))
print(add.__annotations__)
````

## 2. 类型标注不会自动检查运行时输入

标注主要服务于开发者、编辑器和静态检查工具。Python默认仍会执行下面的调用：

````python
print(add("hello ", "world"))

# 虽然标注要求int，但Python没有仅凭标注阻止字符串。
# 如果业务必须在运行时拒绝错误类型，仍要使用isinstance()等检查。
````

## 3. 容器类型

只写`list`或`dict`无法说明内部元素类型。现代Python可以写：

```python
names: list[str] = ["Alice", "Bob"]
scores: dict[str, int] = {"Alice": 90}
point: tuple[int, int] = (10, 20)
tags: set[str] = {"python", "api"}
```

`dict[str, int]`表示键为字符串、值为整数的字典。

````python
def average(scores: list[float]) -> float:
    return sum(scores) / len(scores)


print(average([80.0, 90.0, 100.0]))
````

## 4. 联合类型与`None`

竖线`|`表示一个值可能属于多种类型：

```python
def normalize_id(value: int | str) -> int:
    return int(value)
```

查找函数可能找到用户，也可能找不到：

```python
def find_user(user_id: int) -> dict[str, object] | None:
    ...
```

`| None`提醒调用者必须考虑“没有结果”的情况。

````python
def find_name(user_id: int) -> str | None:
    names = {1: "Alice", 2: "Bob"}
    return names.get(user_id)


name = find_name(1)
if name is None:
    print("没有找到用户")
else:
    print(name.upper())
````

## 5. 变量标注与`object`

变量也可以标注：

```python
port: int = 8000
users: list[dict[str, object]] = []
```

`object`表示可能是任意Python对象。本课的用户字典同时包含整数、字符串和布尔值，所以暂时用`dict[str, object]`。后续会学习更精确的数据模型。
