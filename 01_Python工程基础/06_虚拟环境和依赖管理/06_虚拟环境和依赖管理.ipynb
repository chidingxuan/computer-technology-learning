## 1. 为什么需要虚拟环境

不同项目可能需要不同版本的同一个包：

```text
项目A → 某依赖的旧版本
项目B → 某依赖的新版本
```

如果所有依赖都安装到同一个Python环境中，升级项目B可能破坏项目A。虚拟环境会为项目建立独立的Python依赖目录。

虚拟环境通常不提交Git；仓库的`.gitignore`已经忽略`.venv/`。

## 2. 创建虚拟环境

在PowerShell进入本课目录：

```powershell
cd "D:\\桌面\\学点东西\\codex—help\\01_Python工程基础\\06_虚拟环境和依赖管理"
```

创建名为`.venv`的虚拟环境：

```powershell
python -m venv .venv
```

`python -m venv`表示使用当前Python运行标准库中的`venv`模块。

## 3. 激活与退出

PowerShell激活命令：

```powershell
.\\.venv\\Scripts\\Activate.ps1
```

激活后提示符通常出现`(.venv)`。检查解释器：

```powershell
python -c "import sys; print(sys.executable)"
```

路径应该指向本课目录中的`.venv\\Scripts\\python.exe`。

退出环境：

```powershell
deactivate
```

## 4. 使用`python -m pip`

激活虚拟环境后安装本课依赖：

```powershell
python -m pip install -r requirements.txt
```

查看已安装包：

```powershell
python -m pip list
```

推荐`python -m pip`而不是只写`pip`，因为前者明确使用当前`python`对应的pip，能减少“包安装到了另一个环境”的问题。

## 5. `requirements.txt`

本课文件中写有：

```text
requests
```

它声明项目依赖`requests`。其他人拿到项目后可以执行同一条安装命令复现依赖。

查看某个包的信息：

```powershell
python -m pip show requests
```

真实项目还需要考虑版本范围和锁定文件。本课先掌握环境隔离与依赖声明，后续项目中再逐步提高可复现性。

## 6. Notebook内核也是一个Python解释器

即使终端激活了`.venv`，已经打开的Notebook也不一定自动切换内核。运行下面单元格查看当前Notebook使用的解释器。

````python
import sys

print(sys.executable)
print(sys.version)
````

## 8. 确认虚拟环境没有进入Git

回到仓库根目录执行：

```powershell
git status
```

输出中不应该出现`.venv`里的大量文件。出现时应先修正`.gitignore`，不要提交虚拟环境。
